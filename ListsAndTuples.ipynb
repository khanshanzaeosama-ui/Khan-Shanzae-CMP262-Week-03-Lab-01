## Part 3 – Working with Numbers

```python
scores = [82,73,55,99,100,64]
print(scores)
```

```python

maximum = max(scores)
print("Highest score: ",maximum)

minimum = min(scores)
print("Lowest score: ",minimum)

total = sum(scores)
print("Total score: ",total)

num_score = len(scores)
print("Score length: ",num_score)

avg = sum(scores)/len(scores)
print("Average: ",avg)
```

## Part 6 – Lists vs. Tuples

1. What is a list? A list is able to store many variables in a bracket and is felxible to update, replace variables and remove variables.

2. What is a tuple? A tuple is able to store many variables in a paranthesis, but unlike list it is immutable and cannot change variables.

3. What is one similarity between a list and a tuple? Both can display variables within them.

4. What is the main difference between them? List can change, add, and remove variables, but tuple cannot since it needs to be conveted into list first in order to do so.

5. What does **mutable** mean? it is able to be changed.

6. Which is mutable: a list or a tuple? List, while tuple is immutable

## Part 7 – Challenge: Student Scores

```python
student_scores = [78, 92, 85, 67, 95, 88, 73]


maximum_score = max(student_scores)
print("Highest score: ",maximum_score)

minimum_score = min(student_scores)
print("Lowest score: ",minimum_score)

total_score = sum(student_scores)
print("Total score: ",total_score)

num_stdnt_score = len(student_scores)
print("Score length: ",num_stdnt_score)

avg_score = sum(student_scores)/len(student_scores)
print("Average: ",avg_score)

sl = 0
for score in student_scores:
    if score>= 80:
        sl+=1
        

print("There are",sl,"scores that are of 80 or higher.")
```
